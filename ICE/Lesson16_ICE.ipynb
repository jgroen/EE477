```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# PARAMETERS
# -----------------------------
N = 1000                 # number of received symbols
amplitude = 1.0          # signal amplitude
noise_std = 0.25         # noise standard deviation

# -----------------------------
# QPSK CONstellation
# -----------------------------
constellation = amplitude * np.array([
    1 + 1j,
    -1 + 1j,
    -1 - 1j,
    1 - 1j
])

# Random transmitted symbols
tx = np.random.choice(constellation, N)

# Add AWGN
noise = noise_std * (
    np.random.randn(N) +
    1j * np.random.randn(N)
)

rx = tx + noise

# -----------------------------
# PLOT
# -----------------------------
plt.figure(figsize=(6, 6))

plt.scatter(
    rx.real,
    rx.imag,
    alpha=0.3,
    s=10,
    label='Received Symbols'
)

plt.scatter(
    constellation.real,
    constellation.imag,
    color='red',
    marker='x',
    s=200,
    linewidths=3,
    label='Ideal Symbols'
)

# Decision boundaries
plt.axhline(0, color='k', linestyle='--')
plt.axvline(0, color='k', linestyle='--')

plt.xlabel('In-Phase (I)')
plt.ylabel('Quadrature (Q)')
plt.title(
    f'QPSK Constellation\n'
    f'Amplitude={amplitude}, Noise Std={noise_std}'
)

plt.grid(True)
plt.axis('equal')
plt.legend()

plt.show()
```
