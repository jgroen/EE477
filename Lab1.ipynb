```python
import numpy as np

# Helper function: Convert data into a string of bits
def to_bits(data):
    return ''.join(format(byte, '08b') for byte in data)

# Helper function: Display the first 32 bits
def first_32_bits(bitstream):
    return bitstream[:32]
```

```python
# TODO: Enter a short text message
message = "Hello, digital world!"

# Convert the text into bytes using UTF-8 encoding
text_bytes = message.encode('utf-8')

# Convert bytes into a binary bitstream
text_bits = to_bits(text_bytes)

# Calculate results
num_characters = len(message)
num_bits = len(text_bits)

# Display results
print("TEXT MESSAGE")
print("------------------------------")
print("Message:", message)
print("Number of characters:", num_characters)
print("Number of bits:", num_bits)
print("First 32 bits:", first_32_bits(text_bits))
```

```python
from IPython.display import Audio, display
from google.colab import output
import numpy as np
import base64
import subprocess

# Record approximately 4 seconds of audio
recording = output.eval_js("""
async function recordAudio() {
    const stream = await navigator.mediaDevices.getUserMedia({
        audio: true
    });

    const recorder = new MediaRecorder(stream);
    const chunks = [];

    recorder.ondataavailable = event => {
        if (event.data.size > 0) chunks.push(event.data);
    };

    const stopped = new Promise(resolve => {
        recorder.onstop = resolve;
    });

    recorder.start();
    await new Promise(resolve => setTimeout(resolve, 4000));
    recorder.stop();
    await stopped;

    stream.getTracks().forEach(track => track.stop());

    const blob = new Blob(chunks, {type: recorder.mimeType});
    const reader = new FileReader();

    const data = await new Promise(resolve => {
        reader.onloadend = () => resolve(reader.result);
        reader.onerror = () => resolve(null);
        reader.readAsDataURL(blob);
    });

    return {
        data: data,
        mimeType: recorder.mimeType,
        size: blob.size
    };
}

recordAudio();
""")

# Check that a recording was received
if not recording or not recording["data"]:
    raise RuntimeError("No audio recording was received.")

print("Recording format:", recording["mimeType"])
print("Recording file size:", recording["size"], "bytes")

# Decode recording
audio_data = base64.b64decode(recording["data"].split(",")[1])

if len(audio_data) == 0:
    raise RuntimeError("The recording is empty.")

# Convert to 16-bit, 16 kHz, mono PCM
result = subprocess.run(
    [
        "ffmpeg", "-y",
        "-i", "pipe:0",
        "-f", "s16le",
        "-acodec", "pcm_s16le",
        "-ac", "1",
        "-ar", "16000",
        "pipe:1"
    ],
    input=audio_data,
    capture_output=True
)

if result.returncode != 0:
    raise RuntimeError(result.stderr.decode())

# Convert bytes into audio samples
audio_bytes = result.stdout
audio_samples = np.frombuffer(audio_bytes, dtype=np.int16)

# Verify recording contains actual audio
if len(audio_samples) == 0:
    raise RuntimeError("No audio samples were produced.")

if np.max(np.abs(audio_samples)) == 0:
    raise RuntimeError(
        "Recording contains only zeros. Check microphone permissions and input."
    )

# Calculate results
sample_rate = 16000
duration = len(audio_samples) / sample_rate
num_bits = len(audio_bytes) * 8

# Convert samples into a binary bitstream
audio_bits = to_bits(audio_bytes)

# Display results
print("\nAUDIO")
print("------------------------------")
print("Recording length:", round(duration, 2), "seconds")
print("Sampling rate:", sample_rate, "samples/second")
print("Number of bits:", num_bits)
print("First 32 bits:", first_32_bits(audio_bits))

# Play recording
display(Audio(audio_samples, rate=sample_rate))
```

```python
# Prepare a bitstream for a selected modulation scheme

def prepare_bits(bitstream, bits_per_symbol):
    """
    Groups a bitstream into symbols of equal length.

    If the bitstream length is not evenly divisible by the
    number of bits per symbol, zeros are appended to the end.
    """

    # Make sure the input contains only binary digits
    if not all(bit in "01" for bit in bitstream):
        raise ValueError("Input must contain only 0s and 1s.")

    # Determine how many zeros are needed to complete the final symbol
    padding_bits = (-len(bitstream)) % bits_per_symbol

    # Append zeros if necessary
    # Padding is added to the END of the bitstream so the original
    # data remains unchanged.
    padded_bitstream = bitstream + ("0" * padding_bits)

    # Divide the bitstream into groups, one for each symbol
    symbols = [
        padded_bitstream[i:i + bits_per_symbol]
        for i in range(0, len(padded_bitstream), bits_per_symbol)
    ]

    return symbols, padding_bits


print("Bitstream preparation function is ready.")
```

```python
# MODULATION
# Map the selected bitstream using four modulation schemes

# Number of bits represented by each symbol
modulations = {
    "BPSK": 1,
    "QPSK": 2,
    "8-PSK": 3,
    "16-QAM": 4
}

# Store the results for each modulation scheme
modulated_data = {}

for modulation, bits_per_symbol in modulations.items():

    # Prepare the bitstream and add padding if necessary
    bit_groups, padding_bits = prepare_bits(input_bits, bits_per_symbol)

    # Map bits to complex-valued modulation symbols
    if modulation == "BPSK":
        # 0 -> +1, 1 -> -1
        symbols = np.array([
            1 if b == "0" else -1 for b in bit_groups
        ], dtype=complex)

    elif modulation == "QPSK":
        # Gray-coded mapping:
        # 00 -> ( 1 + 1j), 01 -> (-1 + 1j)
        # 11 -> (-1 - 1j), 10 -> ( 1 - 1j)
        mapping = {
            "00": (1 + 1j),
            "01": (-1 + 1j),
            "11": (-1 - 1j),
            "10": (1 - 1j)
        }
        symbols = np.array([
            mapping[b] / np.sqrt(2) for b in bit_groups
        ])

    elif modulation == "8-PSK":
        # Gray-coded mapping around the constellation:
        # 000, 001, 011, 010, 110, 111, 101, 100
        # Each successive point is separated by 45 degrees.
        gray_order = ["000", "001", "011", "010",
                      "110", "111", "101", "100"]

        mapping = {
            bits: np.exp(1j * 2 * np.pi * k / 8)
            for k, bits in enumerate(gray_order)
        }
        symbols = np.array([mapping[b] for b in bit_groups])

    elif modulation == "16-QAM":
        # Gray-coded mapping on each axis:
        # 00 -> +3, 01 -> +1, 11 -> -1, 10 -> -3
        axis_mapping = {
            "00": 3,
            "01": 1,
            "11": -1,
            "10": -3
        }

        symbols = np.array([
            (axis_mapping[b[:2]] + 1j * axis_mapping[b[2:]]) / np.sqrt(10)
            for b in bit_groups
        ])

    # Save the symbols and padding information
    modulated_data[modulation] = {
        "symbols": symbols,
        "padding": padding_bits,
        "bits_per_symbol": bits_per_symbol
    }

    # Display results
    print("\n" + modulation)
    print("------------------------------")
    print("Padding bits added:", padding_bits)
    print("First 16 symbols:", symbols[:16])
```
