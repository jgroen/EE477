```python
# ============================================================
# Task 1: Examine the Signal
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio, display

# Load the signal
fs, signal = wavfile.read("noise.wav")

# Convert integer samples to floating-point values for analysis
signal = signal.astype(np.float64)

# Normalize for plotting/analysis
signal = signal / np.max(np.abs(signal))

# Time axis
t = np.arange(len(signal)) / fs

# ------------------------------------------------------------
# Play the signal
# ------------------------------------------------------------

print(f"Sampling rate: {fs:,} Hz")
print(f"Duration: {len(signal) / fs:.3f} seconds")
print(f"Number of samples: {len(signal):,}")

display(Audio(signal, rate=fs))

# ------------------------------------------------------------
# Time-domain representation
# ------------------------------------------------------------

plt.figure(figsize=(12, 4))
plt.plot(t, signal)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Signal in the Time Domain")
plt.grid(True)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Frequency-domain representation
# ------------------------------------------------------------

# Compute the one-sided magnitude spectrum
N = len(signal)
spectrum = np.fft.rfft(signal)
frequencies = np.fft.rfftfreq(N, d=1/fs)

magnitude = np.abs(spectrum) / N
magnitude[1:-1] *= 2

# Convert to dB relative to the largest spectral component
magnitude_db = 20 * np.log10(magnitude / np.max(magnitude) + 1e-12)

plt.figure(figsize=(12, 4))
plt.plot(frequencies, magnitude_db)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB, relative)")
plt.title("Provided Signal Spectrum")
plt.xlim(0, fs / 2)
plt.ylim(-60, 5)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Task 2: Compare Sampling Strategies
# ============================================================

from scipy.signal import resample_poly

# Provided sampling rates
sampling_rates = [44000, 16000, 8000]

print("Sampling rates to compare:")
for rate in sampling_rates:
    print(f"  {rate:,} Hz  (Nyquist frequency = {rate/2:,.0f} Hz)")


def resample_signal(signal, original_fs, new_fs):
    """Resample a signal to a new sampling rate."""
    # Find integer factors for polyphase resampling
    from math import gcd

    g = gcd(original_fs, new_fs)
    up = new_fs // g
    down = original_fs // g

    return resample_poly(signal, up, down)


# Generate, play, and plot each sampled version
fig, axes = plt.subplots(len(sampling_rates), 1, figsize=(12, 10))

for ax, new_fs in zip(axes, sampling_rates):

    sampled_signal = resample_signal(signal, fs, new_fs)

    # Play the sampled signal
    print(f"\nSampling rate: {new_fs:,} Hz")
    display(Audio(sampled_signal, rate=new_fs))

    # Show 10 ms starting about 2/3 of the way through the signal
    duration_to_plot = 0.010
    start_time = (len(signal) / fs) * (2/3)
    start_sample = int(start_time * new_fs)
    n_plot = int(duration_to_plot * new_fs)

    t_sampled = np.arange(n_plot) / new_fs + start_time

    ax.plot(
        t_sampled * 1000,
        sampled_signal[start_sample:start_sample + n_plot],
        marker='o',
        markersize=3,
        linestyle='-'
    )

    ax.set_title(
        f"Sampling Rate = {new_fs:,} Hz "
        f"(Nyquist = {new_fs/2:,.0f} Hz)"
    )
    ax.set_xlabel("Time (ms)")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Frequency-domain comparison
# ------------------------------------------------------------

fig, axes = plt.subplots(len(sampling_rates), 1, figsize=(12, 10))

for ax, new_fs in zip(axes, sampling_rates):

    sampled_signal = resample_signal(signal, fs, new_fs)

    # Compute one-sided FFT
    N = len(sampled_signal)
    spectrum = np.fft.rfft(sampled_signal)
    frequencies = np.fft.rfftfreq(N, d=1/new_fs)

    magnitude = np.abs(spectrum) / N
    magnitude[1:-1] *= 2

    # Convert to dB relative to the maximum
    magnitude_db = 20 * np.log10(
        magnitude / np.max(magnitude) + 1e-12
    )

    ax.plot(frequencies / 1000, magnitude_db)

    ax.set_title(
        f"Sampling Rate = {new_fs:,} Hz "
        f"(Nyquist = {new_fs/2:,.0f} Hz)"
    )
    ax.set_xlabel("Frequency (kHz)")
    ax.set_ylabel("Magnitude (dB, relative)")
    ax.set_xlim(0, fs / 2 / 1000)
    ax.set_ylim(-60, 5)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Task 5: Compression and Recommendation
# ============================================================

import gzip

# ------------------------------------------------------------
# Provided Code
# ------------------------------------------------------------

def quantize_signal(signal, levels):
    """Quantize a normalized signal to a specified number of levels."""

    step = 2 / levels

    quantized = np.round((signal + 1) / step) * step - 1
    quantized = np.clip(quantized, -1, 1)

    return quantized


def compress_signal(quantized_signal):
    """Convert a 256-level quantized signal to bytes and compress it."""

    # Represent each quantized sample using 8 bits (1 byte)
    signal_uint8 = np.round(
        (quantized_signal + 1) * 127.5
    ).astype(np.uint8)

    # Compress the byte representation
    compressed_data = gzip.compress(signal_uint8.tobytes())

    return compressed_data


# ------------------------------------------------------------
# Quantize the Original Signal
# ------------------------------------------------------------

quantization_levels = 256

quantized_signal = quantize_signal(
    signal,
    quantization_levels
)


# ------------------------------------------------------------
# Compress the Quantized Signal
# ------------------------------------------------------------

compressed_data = compress_signal(quantized_signal)


# ------------------------------------------------------------
# Calculate Representation Sizes
# ------------------------------------------------------------

sample_count = len(signal)
sample_rate = fs

uncompressed_bits = sample_count * 8
compressed_bits = len(compressed_data) * 8

uncompressed_kbits = uncompressed_bits / 1000
compressed_kbits = compressed_bits / 1000

compression_ratio = uncompressed_bits / compressed_bits


# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print(f"Original sample rate: {sample_rate} samples/s")
print(f"Number of samples: {sample_count}")
print(f"Quantization levels: {quantization_levels}")
print(f"Uncompressed representation: {uncompressed_kbits:.1f} kbits")
print(f"Compressed representation: {compressed_kbits:.1f} kbits")
print(f"Compression ratio: {compression_ratio:.2f}:1")
```
